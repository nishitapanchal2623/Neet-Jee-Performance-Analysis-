## Phase 2: Data Cleaning

### Objective

The objective of this notebook is to assess the quality of the dataset and perform the necessary preprocessing steps before exploratory data analysis. We will identify and handle missing values, duplicate records, inconsistent data types, and invalid values while preserving the integrity of the original dataset.

```python
# Import Required Libraries
import pandas as pd
import numpy as np

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

## Load the Dataset

The original dataset is loaded from the raw data folder. All preprocessing will be performed on a copy of the dataset to ensure that the raw data remains unchanged.

```python
# Load Dataset
df = pd.read_csv("../data/raw/College_Admission.csv")

# Create a copy for cleaning
clean_df = df.copy()

print("Dataset loaded successfully!")
print(f"Rows: {clean_df.shape[0]}")
print(f"Columns: {clean_df.shape[1]}")
```

### Observation

A copy of the original dataset has been created for preprocessing. This ensures that the raw dataset remains unchanged and supports reproducible data analysis.

## Missing Value Analysis

The first step in data cleaning is to identify missing values. Missing data can affect statistical analysis and machine learning models, so it is important to evaluate whether any preprocessing is required.

```python
# Missing Values

missing_values = clean_df.isnull().sum()

missing_values
```

```python
# Total Missing Values

missing_values.sum()
```

### Observation

The dataset contains **0** missing values.

Based on the results, further preprocessing decisions will be taken if necessary.

## Duplicate Record Analysis

Duplicate records can introduce bias into statistical analysis and machine learning models. Therefore, the dataset is examined to identify whether duplicate observations are present.

```python
# Check Duplicate Records

duplicate_count = clean_df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

### Observation

The dataset contains **0 duplicate records**.

Appropriate preprocessing decisions will be taken based on the result.

## Data Type Validation

Data type validation ensures that each feature is stored in the appropriate format for analysis. Correct data types improve computational efficiency and prevent errors during preprocessing and visualization.

```python
# Check Data Types

clean_df.dtypes
```

### Observation

The dataset contains both numerical and categorical variables. The data types are appropriate for the current analysis, and no data type conversion is required at this stage.

```python
# Check unique values in categorical columns

categorical_columns = clean_df.select_dtypes(include="object").columns

for column in categorical_columns:
    print("=" * 50)
    print(f"Column: {column}")
    print(clean_df[column].unique())
    print()
```

### Observation

The categorical variables were examined for consistency and uniqueness. All categories are standardized without spelling inconsistencies, formatting errors, or duplicate category names. Therefore, no categorical value normalization is required before further analysis.

## Data Validation

Before performing exploratory data analysis, numerical variables are validated to ensure that all values fall within logical and acceptable ranges. This step helps identify invalid or unrealistic observations that may negatively impact statistical analysis.

```python
# Check for Invalid Values

print("Invalid Age:", clean_df[clean_df["age"] < 17].shape[0])

print("Invalid Board Percentage:",
      clean_df[(clean_df["board_percentage"] < 0) |
               (clean_df["board_percentage"] > 100)].shape[0])

print("Invalid Entrance Score:",
      clean_df[clean_df["entrance_score"] < 0].shape[0])

print("Invalid Admission Probability:",
      clean_df[(clean_df["admission_probability"] < 0) |
               (clean_df["admission_probability"] > 1)].shape[0])

print("Invalid Extracurricular Score:",
      clean_df[(clean_df["extracurricular_score"] < 0) |
               (clean_df["extracurricular_score"] > 10)].shape[0])
```

### Observation

The numerical variables were validated against predefined logical ranges. No invalid values were detected for age, board percentage, entrance score, admission probability, or extracurricular score. Therefore, the dataset satisfies the required quality standards for further analysis.

```python
# Save Clean Dataset

clean_df.to_csv("../data/cleaned/College_Admission_Cleaned.csv",
                index=False)

print("Clean dataset saved successfully!")
```
