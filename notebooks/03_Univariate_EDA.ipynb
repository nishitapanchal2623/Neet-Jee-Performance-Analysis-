## Phase 3: Univariate Exploratory Data Analysis (EDA)

### Objective

The objective of this notebook is to analyze each variable independently to understand its distribution, frequency, variability, and overall characteristics. Univariate analysis provides valuable insights into individual features before exploring relationships between variables.

```python
#Import Required Libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

# Plot Style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully!")
```

## Load the Clean Dataset

The cleaned dataset generated during the data preprocessing phase is loaded for exploratory data analysis. Using the cleaned dataset ensures that the analysis is performed on validated and high-quality data.

```python
## Load the Clean Dataset
df = pd.read_csv("../data/cleaned/College_Admission_Cleaned.csv")

print("Dataset Loaded Successfully!")
print(f"Rows : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

### Observation

The cleaned dataset has been loaded successfully and is ready for exploratory data analysis. Since the dataset has already been validated during the data cleaning phase, all subsequent visualizations and statistical analyses will be performed on this dataset.

# Numerical Variable Analysis

The numerical variables are analyzed individually to understand their distribution, central tendency, spread, and potential outliers. Histograms and box plots are used to visualize the characteristics of each feature.

## Distribution of Student Age

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="age",
    bins=4,
    kde=True,
    color="steelblue",
    edgecolor="black"
)

plt.title("Distribution of Student Age", fontsize=16, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Number of Students")

plt.tight_layout()

plt.show()
```

### Observation

The age distribution indicates that the dataset includes students aged between **17 and 20 years**. The frequency of students is nearly uniform across all age groups, suggesting a well-balanced dataset with no age-related bias. No outliers or abnormal age values are observed, confirming that the age feature is suitable for further exploratory analysis.

## Distribution of Entrance Examination Score

The mean entrance score (77.88) is substantially higher than the median score (20), further confirming the presence of a right-skewed distribution.

# Categorical Variable Analysis

The student category distribution is well balanced across all reservation groups, including General, EWS, SC, OBC, and ST. Each category contains approximately 5,000 students, with only minor differences in frequency. This balanced representation reduces category-related sampling bias and supports fair comparative analysis throughout the project.

## Entrance Examination-wise Performance

This analysis compares the average entrance examination score across JEE, NEET, and CET candidates to identify differences in examination performance.

```python
exam_performance = (
    df[df["entrance_exam"].isin(["jee", "neet", "cet"])]
    .groupby("entrance_exam")["entrance_score"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

exam_performance
```

```python
plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=exam_performance.reset_index(),
    x="entrance_exam",
    y="mean",
    hue="entrance_exam",
    palette="Set2",
    legend=False
)

plt.title("Average Entrance Score by Examination", fontsize=16, fontweight="bold")
plt.xlabel("Entrance Examination")
plt.ylabel("Average Entrance Score")

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f")

plt.tight_layout()

plt.savefig(
    "../figures/average_entrance_score_by_exam.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Observation

The average entrance score differs considerably across the three examination categories. NEET has the highest mean score (354.78), followed by JEE (146.50) and CET (99.97). The median scores show the same ordering. NEET also has the highest standard deviation (201.65), indicating greater variation in its recorded scores. However, direct comparison of raw scores across examinations should be interpreted cautiously because different examinations may use different scoring scales.

## Admission Probability by Entrance Examination

```python
exam_admission = (
    df[df["entrance_exam"].isin(["jee", "neet", "cet"])]
    .groupby("entrance_exam")["admission_probability"]
    .agg(["count", "mean", "median", "std"])
    .round(3)
)

exam_admission
```

```python
plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=exam_admission.reset_index(),
    x="entrance_exam",
    y="mean",
    hue="entrance_exam",
    palette="Set2",
    legend=False
)

plt.title("Average Admission Probability by Examination",
          fontsize=16, fontweight="bold")
plt.xlabel("Entrance Examination")
plt.ylabel("Average Admission Probability")

for container in ax.containers:
    ax.bar_label(container, fmt="%.3f")

plt.tight_layout()

plt.savefig(
    "../figures/average_admission_probability_by_exam.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Observation

The average admission probability varies across entrance examination categories. NEET has the highest average admission probability of 0.505, followed by JEE at 0.372 and CET at 0.339. The median values follow the same pattern. Within this dataset, NEET candidates therefore have the highest recorded admission probability on average, while CET candidates have the lowest. These results indicate an association between entrance examination category and admission probability, but they do not establish a causal relationship.

## Relationship Between Entrance Score and Admission Probability

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="entrance_score",
    y="admission_probability",
    alpha=0.4
)

sns.regplot(
    data=df,
    x="entrance_score",
    y="admission_probability",
    scatter=False
)

plt.title(
    "Entrance Score vs Admission Probability",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Entrance Score")
plt.ylabel("Admission Probability")

plt.tight_layout()

plt.savefig(
    "../figures/entrance_score_vs_admission_probability.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
correlation = df["entrance_score"].corr(
    df["admission_probability"]
)

print("Correlation:", round(correlation, 3))
```

### Observation

The scatter plot shows a positive relationship between entrance score and admission probability. The regression line indicates that admission probability generally increases as entrance score increases. The Pearson correlation coefficient is 0.554, indicating a moderate positive linear relationship between the two variables. Therefore, higher entrance scores are generally associated with higher admission probabilities in this dataset.

## State-wise Entrance Score Performance

This analysis compares the average entrance examination score across states to identify geographical differences in recorded student performance.

```python
state_performance = (
    df.groupby("state")["entrance_score"]
      .agg(["count", "mean", "median", "std"])
      .sort_values("mean", ascending=False)
      .round(2)
)

state_performance
```

```python
plt.figure(figsize=(12, 8))

ax = sns.barplot(
    data=state_performance.reset_index(),
    x="mean",
    y="state",
    hue="state",
    palette="viridis",
    legend=False
)

plt.title(
    "Average Entrance Score by State",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Average Entrance Score")
plt.ylabel("State")

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f", padding=3)

plt.tight_layout()

plt.savefig(
    "../figures/state_wise_average_entrance_score.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Observation

The average entrance score varies across states. Andhra Pradesh has the highest average entrance score at 85.68, followed by Goa at 83.87 and Mizoram at 82.67. Telangana has the lowest average at 72.38, followed by Uttar Pradesh at 73.09. The difference between the highest and lowest state averages is approximately 13.30 points. Since the number of students represented by each state is relatively similar, the state-wise comparison provides a useful overview of geographical variation in recorded entrance scores. These differences indicate variation within the dataset but should not be interpreted as evidence that students from one state inherently perform better than another.
